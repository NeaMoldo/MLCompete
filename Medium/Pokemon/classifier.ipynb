```python
import torch
from torchvision import transforms as T
from torch.utils.data import DataLoader, Dataset, Subset
import torch.nn as nn
import torch.optim as optim
import torchvision.models as models
import torchmetrics

import pytorch_lightning as L
from pytorch_lightning.callbacks import EarlyStopping, ModelCheckpoint
from sklearn.model_selection import train_test_split

import numpy as np
import pandas as pd
from PIL import Image
```

```python
class ImageDataset(Dataset):
    def __init__(self, csv_file, transform=None, is_test=False):
        self.df = pd.read_csv(csv_file)
        self.transform = transform
        self.is_test = is_test

        if not self.is_test:
            self.classes = sorted(self.df['Type'].unique())
            self.class_to_idx = {cls_name: i for i, cls_name in enumerate(self.classes)}
            self.idx_to_class = {i: cls_name for i, cls_name in enumerate(self.classes)}

    def __len__(self):
        return len(self.df)

    def __getitem__(self,idx):
        row = self.df.iloc[idx]
        image = Image.open(row['ImagePath']).convert('RGB')
        img_id = row['SampleID']

        if self.transform:
            image = self.transform(image)

        if self.is_test:
            return image, img_id
        else:
            label = self.class_to_idx[row['Type']]
            return image, torch.tensor(label, dtype=torch.long)


class DataModule(L.LightningDataModule):
    def __init__(
        self, train_csv, test_csv, batch_size=32, img_size=224, val_split=0.2
    ):
        super().__init__()
        self.train_csv = train_csv
        self.test_csv = test_csv
        self.batch_size =  batch_size
        self.img_size = img_size
        self.val_split = val_split

        self.transform_train = T.Compose([
            T.Resize((img_size,img_size)),
            T.RandomHorizontalFlip(),
            T.RandomRotation(15),
            T.ColorJitter(
                brightness=0.2, contrast=0.2, saturation=0.2
            ),
            T.ToTensor(),
            T.Normalize(
                mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]
            )
        ])

        self.transform_val = T.Compose([
            T.Resize((img_size,img_size)),
            T.ToTensor(),
            T.Normalize(
                mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]
            )
        ])

    def setup(self, stage=None):
        full_dataset = ImageDataset(self.train_csv, transform=self.transform_train)
        self.num_classes = len(full_dataset.classes)
        self.class_to_idx = full_dataset.class_to_idx
        self.idx_to_class = full_dataset.idx_to_class

        train_idx, val_idx = train_test_split(
            range(len(full_dataset)),
            test_size=self.val_split,
            random_state=42,
            stratify=full_dataset.df['Type']
        )

        self.train_dataset = Subset(full_dataset, train_idx)

        val_dataset_eval = ImageDataset(self.train_csv, transform=self.transform_val)
        self.val_dataset = Subset(val_dataset_eval, val_idx)
        self.test_dataset = ImageDataset(self.test_csv, transform=self.transform_val, is_test=True)

    def train_dataloader(self):
        return DataLoader(
            self.train_dataset,
            batch_size=self.batch_size,
            shuffle=True,
            num_workers=4,
            persistent_workers=True
        )

    def val_dataloader(self):
        return DataLoader(
            self.val_dataset,
            batch_size=self.batch_size,
            shuffle=False,
            num_workers=4,
            persistent_workers=True
        )

    def test_dataloader(self):
        return DataLoader(
            self.test_dataset,
            batch_size=self.batch_size,
            shuffle=False,
            num_workers=4,
            persistent_workers=True
        )
    
```

```python
class ClassifierModule(L.LightningModule):
    def __init__(self, num_classes, lr=1e-3):
        super().__init__()
        self.save_hyperparameters()

        self.backbone = models.resnet34(weights=models.ResNet34_Weights.DEFAULT)
        self.backbone.fc = nn.Linear(self.backbone.fc.in_features, num_classes)

        self.criterion = nn.CrossEntropyLoss()

        self.train_acc = torchmetrics.Accuracy(
            task='multiclass', num_classes=num_classes
        )

        self.val_acc = torchmetrics.Accuracy(
            task='multiclass', num_classes=num_classes
        )

    def forward(self, x):
        return self.backbone(x)

    def training_step(self, batch, batch_idx):
        x, y = batch
        logits = self(x)
        loss = self.criterion(logits, y)
        preds = torch.argmax(logits, dim=1)
        self.train_acc(preds, y)

        self.log('train_loss', loss, on_epoch=True, on_step=True, prog_bar=True, logger=True)
        self.log('train_acc', self.train_acc, on_step=False, on_epoch=True, prog_bar=True, logger=True)

    def validation_step(self, batch, batch_idx):
        x, y = batch
        logits = self(x)
        loss = self.criterion(logits, y)
        preds = torch.argmax(logits, dim=1)
        self.val_acc(logits, y)

        self.log('val_loss', loss, on_epoch=True, on_step=False, prog_bar=True, logger=True)
        self.log('val_acc', self.val_acc, on_epoch=True, on_step=False, prog_bar=True, logger=True)

    def configure_optimizers(self):
        opt = optim.AdamW(self.parameters(), lr = self.hparams.lr, weight_decay=1e-4)
        sch = optim.lr_scheduler.ReduceLROnPlateau(optimizer=opt, mode='min', factor=0.1, patience=2)

        return {
            'optimizer': opt,
            'lr_scheduler': {
                'scheduler': sch,
                "monitor": 'val_loss'
            }
        }
```

```python
dm = DataModule(
    train_csv='train.csv', test_csv='test.csv', batch_size=32, img_size=224
)
dm.setup()

model = ClassifierModule(num_classes=dm.num_classes, lr=1e-3)

checkpoint_callback = ModelCheckpoint(
    monitor='val_loss',
    dirpath='checkpoints/',
    filename='pokemon-resnet-{epoch:02d}-{val_loss:.4f}',
    save_top_k=1,
    mode='min'
)

early_stop_callback = EarlyStopping(
    monitor='val_loss', patience=5, mode='min', verbose=True
)

trainer = L.Trainer(
    max_epochs=20,
    accelerator='auto',
    devices=1,
    callbacks=[checkpoint_callback, early_stop_callback],
    precision='16-mixed'
)

trainer.fit(model, dm)
```

```python
best_model_path = checkpoint_callback.best_model_path
model = ClassifierModule.load_from_checkpoint(best_model_path)

model.eval()
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

test_loader = dm.test_dataloader()
results = []

with torch.no_grad():
    for images, img_ids in test_loader:
        images = images.to(device)
        outputs = model(images)
        preds = torch.argmax(outputs, dim=1).cpu().numpy()

        for img_id, pred_idx in zip(img_ids, preds):
            label_name = dm.idx_to_class[pred_idx]
            sample_id = img_id.item()
            results.append({
                'SampleID': sample_id,
                'Type': label_name
            })

submission_df = pd.DataFrame(results)
submission_df.to_csv('submission.csv', index=False)
```
